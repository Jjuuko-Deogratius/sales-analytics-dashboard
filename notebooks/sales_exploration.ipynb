Exploratory Data Analysis & Visualization

```python
import pandas as pd
import plotly.express as px
```

Question 1 — Which branch generates the most revenue?

```python
branch_revenue = (
    df.groupby("Branch")["Total_Sales"]
    .sum()
    .reset_index()
)

branch_revenue
```

Now sort it:

```python
branch_revenue = branch_revenue.sort_values(
    "Total_Sales",
    ascending=False
)

branch_revenue
```

Create the chart

```python
fig = px.bar(
    branch_revenue,
    x="Branch",
    y="Total_Sales",
    title="Revenue by Branch",
    labels={"Total_Sales": "Revenue (UGX)"}
)

fig.show()
```

Question 2 — Which categories generate the most revenue?

```python
category_revenue = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

category_revenue
```

```python
fig = px.bar(
    category_revenue,
    x="Category",
    y="Total_Sales",
    title="Revenue by Product Category",
    labels={"Total_Sales": "Revenue (UGX)"}
)

fig.show()
```

Question 3 — Which products generate the most revenue?

```python
product_revenue = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

product_revenue.head()
```

```python
top_5_products = product_revenue.head(5)

top_5_products
```

```python
fig = px.bar(
    top_5_products,
    x="Product",
    y="Total_Sales",
    title="Top 5 Products by Revenue",
    labels={"Total_Sales": "Revenue (UGX)"}
)

fig.show()
```

Revenue and popularity are different  
 
This is an important business concept.  

The product generating the most money may not be the product selling the most units.

Let's examine quantity separately.

```python
product_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .reset_index()
    .sort_values("Quantity", ascending=False)
)

product_quantity.head()
```

```python
top_5_quantity = product_quantity.head(5)
```

```python
fig = px.bar(
    top_5_quantity,
    x="Product",
    y="Quantity",
    title="Top 5 Products by Units Sold",
    labels={"Quantity": "Units Sold"}
)

fig.show()
```

Question 4 — How is revenue changing over time?

```python
monthly_revenue = (
    df.groupby("Year_Month")["Total_Sales"]
    .sum()
    .reset_index()
)

monthly_revenue
```

```python
fig = px.line(
    monthly_revenue,
    x="Year_Month",
    y="Total_Sales",
    markers=True,
    title="Monthly Revenue Trend",
    labels={
        "Year_Month": "Month",
        "Total_Sales": "Revenue (UGX)"
    }
)

fig.show()
```

Question 5 — How do customers pay?

Question 8 — How do branches perform by category?

Does every branch earn its revenue from the same product categories?

```python
branch_category_revenue = (
    df.groupby(["Branch", "Category"])["Total_Sales"]
    .sum()
    .reset_index()
)

branch_category_revenue
```

```python
fig = px.bar(
    branch_category_revenue,
    x="Branch",
    y="Total_Sales",
    color="Category",
    title="Revenue by Branch and Category",
    labels={"Total_Sales": "Revenue (UGX)"}
)

fig.show()
```

Average transaction value by branch  

Which branch has customers spending more per transaction?  

```python
branch_average = (
    df.groupby("Branch")["Total_Sales"]
    .mean()
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

branch_average
```

```python
fig = px.bar(
    branch_average,
    x="Branch",
    y="Total_Sales",
    title="Average Transaction Value by Branch",
    labels={"Total_Sales": "Average Transaction Value (UGX)"}
)

fig.show()
```

Compare transactions by branch

```python
branch_transactions = (
    df.groupby("Branch")["Transaction_ID"]
    .nunique()
    .reset_index()
)

branch_transactions.columns = [
    "Branch",
    "Transactions"
]

branch_transactions
```

```python
fig = px.bar(
    branch_transactions,
    x="Branch",
    y="Transactions",
    title="Number of Transactions by Branch"
)

fig.show()
```

Create our KPI summary

```python
total_revenue = df["Total_Sales"].sum()

total_transactions = df["Transaction_ID"].nunique()

total_units = df["Quantity"].sum()

average_transaction_value = (
    total_revenue / total_transactions
)
```

```python
print(f"Total Revenue: UGX {total_revenue:,.0f}")
print(f"Total Transactions: {total_transactions:,}")
print(f"Total Units Sold: {total_units:,}")
print(
    f"Average Transaction Value: "
    f"UGX {average_transaction_value:,.0f}"
)
```

useful EDA summary table

```python
product_summary = (
    df.groupby(["Product", "Category"])
    .agg(
        Units_Sold=("Quantity", "sum"),
        Revenue=("Total_Sales", "sum"),
        Transactions=("Transaction_ID", "nunique")
    )
    .reset_index()
)

product_summary.head()
```

```python
product_summary = product_summary.sort_values(
    "Revenue",
    ascending=False
)

product_summary
```
